# Question 2 (10 marks)

Write a few sentences to describe the system that you are going to model.

---

The system models the post afl match transit and pedestrian flow dynamics of rail-bound fans exiting the Melbourne Cricket Ground (MCG) over the 90-minute window following the final siren. I will measure the impact of changing train capacity (mostly by increasing frequency), and the geographic demographics of fans of the teams playing, on the proportion of fans successfully going home by the end of this window.

## **Bounds** of the system:
- The pedestrian and rail flow of the system will be bounded by the three rail hubs immediately surrounding the MCG precinct and their platform capacities, which are Flinders St/Town Hall stations (15 platforms), Richmond station (10 platforms), and Jolimont station (2 platforms). The pedestrian flow will therefore be restricted to Yarra Park and Birrarung Marr, both of which can be assumed to be very high capacity with limited bottlenecking effects compared to those associated with train capacity directly.
- Total match attendance will be broadly partitioned into the three stations based on team supporter demographics. Values will be based on publicly available average attendance data, while team support demographics can be based on both historical AFL heartlands if no accessible public data is available. Teams will be limited to those with large Victorian supporter bases (Victorian based teams + Brisbane, Sydney), with interstate teams not considered. Potentially, if I can find the data to support it, I can also add factors for increased attendance during rivalry games.
- I also note the existence of the idea that passengers may be more likely to travel to a more centralised station if queue times are excessive, based on anecdotal evidence, but the actual impact of this will be taken to be negligible for the structure of this model on the assumption that Melbourne's radial rail network means that there are limited alternate options for going home other than the main rail lines for each demographic region. (Eg. The only way to get to Mernda is via Jolimont station, and going to Richmond or Flinders will not change this).
- On this note, we assume some effective maximum station capacity. This is realistic since there can only have a limited number of people (alive) in each station, but in reality we can also assume people wait outside the gates while still queueing. We will consider these people outside of the station and part of the inflow rate.

## System dynamics and **feedbacks**:
- Platform congestion modelling. Once passenger platform density hits [critical values](https://kaiwave.allez.la/index.php/s/go3ynXKt6s9CKzF) (I will admit I haven't read this document, but it seems like it may useful for future reference), doorway congestion and other safety measures will train waiting times and boarding efficiency, reducing the effective outflow and increasing rates of queue growth.
- Pedestrian surge redistribution. Excessive station dwelling times or queue lengths may result in a higher proportion of pedestrians taking transit out later, which can 'flatten the curve' and distribute pedestrian surges further out in time. 
- Walking distance to rail hubs. The immediate proximity of Richmond and Jolimont stations as opposed to flinders street may result in much more condensed, higher passenger surges, especially during games with high geographic bias towards these regions.
- Most of these dynamics are based on differential systems where values affect rates of change and vice versa, so I expect nice levels of complexity! Elaborating on this, the first two points are the main source for dynamic feedback, since rates of change are involved, it is necessary to calculate not only values at some current $t$, but also derivatives associated with the state of the current system at $t$ in order to calculate the next step.

# Question 3 (10 marks)

List the variables involved – identify whether they are independent variables or dependent variables.

---

## Independent Variables
- Total attendance: Total number of fans exiting the stadium ($Pax$)
- Demographic origin splits: Proportions of the crowd travelling to North, East/South-East, and Western/Regional corridors based on the competing teams (No units)
- Train frequency: Departing trains per unit $t$ allocated to each station (No units)
- Train capacity: Rated passenger capacity of each departing train (No units)
- (Baseline station capacity): Simply frequency * train capacity * platform amounts ($Pax/Min$)

## Dependent Variables
- Station inflow rates: The number of people arriving at each station per unit $t$ ($Pax/Min$)
- Queue sizes: Backlog of waiting passengers at each moment $t$ ($Pax$)
- Station outflow rates: The number of people departing at each station per unit $t$. The effective station capacity after outflow adjustments are made. ($Pax/Min$)
- Proportion cleared: The number of people cleared after 90 minutes ($Pax$)

## Memory/Function Variables
- Non transit crowd number: The amount of fans who have departed the venue but are not going to the station (inflow) or queueing at platforms. Conserves total crowd values when inflow rates are effected. ($Pax$)
- Internal platform size: Physical size of each platform/effective platform size per station ($m^2$)
- Crowd density: Related to platform size, used for the function which reduces boarding efficiency in high crowding environments ($Pax/m^2$)

# Question 6 (25 marks)

Describe some of the complex relationships that you can deduct from your conceptual model.

---

Most of the calculations which occur in my model are relationships between inflow/outflow and queue times. As stated earlier, the fact that this is a differential system as opposed to some standard algebraic system introduces some unique dynamic complexities inherently associated with rates of change functions. Also, since this is essentially three systems connected to two independent variables each, one of which are coupled (the fan attendence proportions), this introduces some fun relationships between each of the outflow rates.

We observe reinforcing (positive feedback) loops between queues and outflow rates. Since queue lengths don't affect outflow in a linear way (queue lengths basically have no effect on outflow rates until it reaches some critical point where it scales like crazy), this should have some interesting effects on the total clearance time ([Source](https://onlinepubs.trb.org/Onlinepubs/trr/1992/1361/1361-041.pdf), again, haven't read it fully but skimmed this information). This means that in a maximally efficient theoretical system, the only effect on outflow rates are the raw capacity variables, but in busy scenarios such as the one we model, this efficiency decrease should have some measurable impact on the speed of which fans can be cleared from each station. However, outflow rates have a clean relationship to queue lengths, being directly proportional to their reduction.

We also observe balancing loops between inflow rates and queue sizes. While inflow directly increases queue lengths, passenger behaviour suggests that queue lengths will affect the inflow rate, since passengers may defer travel to a later time to avoid queues. This suggests that in some manner, both outflow and inflow rates will affect queue rates, which affect each other. This will certainly be a complex relationship to model. 

All of inflow and outflow rates are directly influenced independently by the initial station capacity values and match demographic weighting, and so even the same match attendance with the same station capacities can see wildly varying results based on the geographic biases of each team! This is probably the most interesting relationship for me personally.
