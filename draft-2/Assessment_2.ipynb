### My answer
#### Formulation
Establish the corridors as

- **1**: Jolimont ($J$)
  
- **2**: Richmond ($R$)

- **3**: Flinders ($F$)

Let $i\in\{J,R,F \}$. For each $i$, define state variables over $t\in[0,90]$ to be
- $M_i(t)$: the MCG remaining pedestrian surge headed towards $i$
- $Q_i(t)$: the platform and immediate queue at $i$
- $C_i(t)$: the numbers cleared after time $t$

All are measured in $n people (pax)$

#### Governing ODEs
Define
- $\frac{dM_i}{dt}=-I_i(t)$
- $\frac{dQ_i}{dt}=I_i(t)-O_i(t)$
- $\frac{dC_i}{dt}=O_i(t)$

With $I,O$ as in and outflow equations (which will be defined later). Then, the total cleared after 90 is
$\sum\limits_{i}C_i(90)$

### Inflow dynamics
Stadium attendance is partitioned by dimensionless demographic corridor proportions $\alpha_i$ where $\sum\limits_i \alpha_i=1$, and $M_i(0)=\alpha_i \cdot N_{total}$. 

And, the discharge follows a gamma surge curve peak (1) . Define
$$f_i(t) = \frac{t^{k-1} e^{-t / \theta_i}}{\theta_i^k \, \Gamma(k)}, \quad t \ge 0$$

Where we have shape k, $\tau_i$ is the walking time $min$, $\Gamma$ is the standard gamma function, and $\theta_i$ is the station specific scale (dimensionless). From the definition of gamma distribution, the time of peak arrival occurs at $(k-1)\theta_i=\tau_i((k-1) /k)$

To capture pedestrian surge feedback redistribution

$$I_i(t) = M_i(0) \cdot f_i(t) \cdot \Psi_i(Q_i)$$

In units of $paxm^{-1}$ and

$$\Psi_{i}(Q_{i}) = \frac{1}{1 + \exp\left(\beta_{q} (Q_i - Q_{i,\text{gate crit}})\right)}$$

(Psi is a dimensionless standard logistic function). It has behaviours where
- If $Q_i$ << gate crit, its free flow
- If $Q_i$ = gate crit, the capacity dampens significantly
- If $Q_i$ >> gate crit, the system chokes to 0

### Platform congestion
We have $C_{base,i}=f_{base,i}\cdot Capacity_{train}$

Crowd density $D_i(t)$ is given by $Q_i(t)/A_i$ in units of $paxm^{-2}$. Platform crowding degrades boarding efficiency according to

$$\eta(D_i) = \eta_{\min} + \frac{1 - \eta_{\min}}{1 + \left( \frac{D_i(t)}{D_{\text{crit}}} \right)^\gamma}$$

Where this is the Four Parameter Logistic regression eqn (2), which I liked because it contains an efficiency baseline $1$, a minimum $\eta_{\min}$, $D_{crit}$ which is the inflection point, and $\gamma$ which is a changeable slope. 

Thus, the outflow is given by
$$O_i(t) = min(Q_i(t), C_{base,i}\cdot \eta(D_i))$$

In $paxm^{-1}$

### My code

```python
from math import exp, gamma as gamma_function, isfinite # Import necessary functions from math

# Define a function which validates parameters that need to be positive !!
# Based on my personal tendency to use silly values and make very avoidable errors...
def _require_positive(value: float, 
                      name: str
                     ) -> None:
    if not isfinite(value) or value <= 0: 
        raise ValueError(f"{name} must be a finite positive number") # Throws error if its not finite positive

# Define Gamma surge density at time t
def gamma_density(time: float, 
                  shape: float, 
                  scale: float
                 ) -> float:
    _require_positive(time, "time") # Validate time
    _require_positive(shape, "shape")
    _require_positive(scale, "scale") # Validate given parameters

    # Equation directly implemented from math section
    density: float = time ** (shape - 1) * exp(-time / scale) / (scale**shape * gamma_function(shape))

    return density

# Define Queue throttle given queue (and parameters) (Psi(Q)) in the math section. 
def queue_throttle(queue: float, 
                   gate_capacity: float, 
                   sensitivity: float
                  ) -> float:
    _require_positive(gate_capacity, "gate_capacity")
    _require_positive(sensitivity, "sensitivity") # Validate given parameters

    if queue < 0: # Validate queue size, cant use require positive since queue starts AT zero
        raise ValueError("queue must not be negative")
    
    argument = sensitivity * (queue - gate_capacity) # Pre calculates the argument since positivity will be checked
    
    if argument >= 0:
        throttle: float = exp(-argument) / (1.0 + exp(-argument)) # Prevent overflow 
    else:
        throttle: float = 1.0 / (1.0 + exp(argument)) 

    return throttle

def inflow_rate(time: float,
                queue: float,
                crowd_allocation: float,
                shape: float,
                scale: float,
                gate_capacity: float,
                sensitivity: float
               ) -> float:
    _require_positive(crowd_allocation, "crowd_allocation") #check crowd number makes sense

    inflow: float = crowd_allocation * gamma_density(time, shape, scale) * queue_throttle(queue, gate_capacity, sensitivity)

    return inflow
```
